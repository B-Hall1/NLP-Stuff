```python
class Category:
  BOOKS = "BOOKS"
  CLOTHING = "CLOTHING"

train_x = "I love the book", "this is a great book", "the fit is great", "I love the shoes"
train_y = [Category.BOOKS, Category.BOOKS, Category.CLOTHING, Category.CLOTHING]
```

```python
!pip install spacy
!python -m spacy download en_core_web_md
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(binary=True, ngram_range=(1,2))
train_x_vectors = vectorizer.fit_transform(train_x)

print(vectorizer.get_feature_names_out())

print(train_x_vectors.toarray())
```

```python
from sklearn import svm

clf = svm.SVC(kernel='linear')
clf.fit(train_x_vectors, train_y)
```

```python
test_x = vectorizer.transform(['I like the book'])

clf.predict(test_x)
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

phrase = "reading the books"
words = word_tokenize(phrase)

lemmatized_words = []
for word in words:
  lemmatized_words.append(lemmatizer.lemmatize(word, pos='v'))

' '.join(lemmatized_words)
```


## Stopword Removal




```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
stop_words = stopwords.words('english')

phrase = "Here is an example sentence demonstrating the removal of stopwords"
words = word_tokenize(phrase)

stripped_phrase = []
for word in words:
  if word not in stop_words:
    stripped_phrase.append(word)

' '.join(stripped_phrase)
```

```python
import spacy
import torch

nlp = spacy.load("en_core_web_trf")
doc = nlp("Here is some text to encode.")

#print(doc.tensor.shape
```

```python
class Category:
  BOOKS = "BOOKS"
  BANK = "BANK"

train_x = ["good characters and plot progression", "check out the book", "good story. would recommend", "novel recommendation", "need to make a deposit to the bank", "balance inquiry savings", "save money"]
train_y = [Category.BOOKS, Category.BOOKS, Category.BOOKS, Category.BOOKS, Category.BANK, Category.BANK, Category.BANK]
```

```python
from sklearn import svm
import numpy as np

docs = [nlp(text) for text in train_x]

train_x_vectors = []
for doc in docs:
    if doc.tensor.shape[0] == 0: # Check if the tensor is empty (no tokens for transformer)
        # Fallback to a zero vector if doc.tensor is empty.
        # en_core_web_trf uses a transformer model, typically BERT-base, which has 768 dimensions.
        train_x_vectors.append(np.zeros(768, dtype=np.float32))
    else:
        # Average across tokens (axis 0) and layers (axis 1) to get a single 1D vector per document
        train_x_vectors.append(doc.tensor.mean(axis=(0, 1)))

clf_svm = svm.SVC(kernel='linear')

clf_svm.fit(train_x_vectors, train_y)

test_x = ["check this story out"]
docs = [nlp(text) for text in test_x]

test_x_vectors = []
for doc in docs:
    if doc.tensor.shape[0] == 0:
        test_x_vectors.append(np.zeros(768, dtype=np.float32))
    else:
        test_x_vectors.append(doc.tensor.mean(axis=(0, 1)))

clf_svm.predict(test_x_vectors)
```
